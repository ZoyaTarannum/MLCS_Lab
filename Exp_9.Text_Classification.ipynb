```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Step 1: Create a sample email dataset
data = {
    "Email": [
        "Win a free iPhone now",
        "Claim your prize today",
        "Congratulations you won a lottery",
        "Get cheap loans immediately",
        "Click here to win cash",
        "Earn money from home easily",
        "Meeting scheduled for tomorrow",
        "Please find the project report attached",
        "Can we discuss the assignment",
        "Your class starts at 10 AM",
        "The team meeting is postponed",
        "Please submit your work today",
        "Free gift waiting for you",
        "You have won a cash reward",
        "Let us meet for lunch tomorrow",
        "Please review the attached document"
    ],
    "Label": [
        "Spam", "Spam", "Spam", "Spam",
        "Spam", "Spam", "Not Spam", "Not Spam",
        "Not Spam", "Not Spam", "Not Spam",
        "Not Spam", "Spam", "Spam", "Not Spam",
        "Not Spam"
    ]
}

df = pd.DataFrame(data)

# Step 2: Separate email text and labels
X = df["Email"]
y = df["Label"]

# Step 3: Convert text into numerical features
vectorizer = CountVectorizer()
X_vectorized = vectorizer.fit_transform(X)

# Step 4: Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_vectorized, y, test_size=0.25,
    random_state=42, stratify=y
)

# Step 5: Train the Naive Bayes classifier
model = MultinomialNB()
model.fit(X_train, y_train)

# Step 6: Evaluate the model
y_pred = model.predict(X_test)

print("Email Spam Classification")
print("Accuracy:", round(accuracy_score(y_test, y_pred) * 100, 2), "%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

# Step 7: Classify new emails
new_emails = [
    "Congratulations you won free cash",
    "Please attend the project meeting tomorrow"
]

new_vectors = vectorizer.transform(new_emails)
predictions = model.predict(new_vectors)

print("\nNew Email Predictions:")
for email, prediction in zip(new_emails, predictions):
    print("Email:", email)
    print("Classification:", prediction)
    print()
```
