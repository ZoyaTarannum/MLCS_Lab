```python
    #Experiment 06
# Detect Abnormal URL Patterns using Feature Engineering

import pandas as pd
import re

# Sample URLs
urls = [
    "https://www.google.com",
    "https://www.amazon.com",
    "https://www.wikipedia.org",
    "http://login-secure-account.com",
    "http://192.168.1.10/login",
    "http://free-prize-winner.com",
    "https://www.github.com",
    "http://verify-account@secure-login.com"
]

# Feature extraction
def extract_features(url):
    return {
        "URL": url,
        "Length": len(url),
        "Dots": url.count("."),
        "Hyphens": url.count("-"),
        "Digits": sum(c.isdigit() for c in url),
        "Special_Characters": len(re.findall(r"[@?=&_%]", url)),
        "Has_IP": bool(re.search(r"(\d{1,3}\.){3}\d{1,3}", url)),
        "Has_HTTPS": url.startswith("https")
    }

# Create DataFrame
df = pd.DataFrame([extract_features(url) for url in urls])

# Detect abnormal patterns
df["Abnormal"] = (
    (df["Length"] > 50) |
    (df["Has_IP"]) |
    (df["Hyphens"] > 2) |
    (df["Special_Characters"] > 2) |
    (~df["Has_HTTPS"])
)

# Display results
print(df)

print("\nAbnormal URLs:")
print(df[df["Abnormal"] == True]["URL"])
```
